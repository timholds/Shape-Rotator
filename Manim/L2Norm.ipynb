```python
from manim import *

# Load the Manim Jupyter extension
%load_ext manim

# Optionally configure video display size
config.media_width = "75%"

class MatrixTransformation(LinearTransformationScene):
    def __init__(self, matrix, **kwargs):
        super().__init__(**kwargs)
        self.matrix = matrix

    def apply_matrix(self, mobject):
        self.apply_matrix_to_points(self.matrix, mobject)

    def apply_matrix_to_points(self, matrix, mobject):
        mobject.apply_matrix(matrix)
        return mobject

class L2NormInducedScene(Scene):
    def construct(self):
        unit_circle = Circle(radius=1, color=BLUE).shift(ORIGIN)
        unit_label = Text("Unit Sphere (L2 norm = 1)", font_size=24).next_to(unit_circle, UP)
        self.play(Create(unit_circle), Write(unit_label))
        self.wait(1)

        initial_vector = np.array([0.7, 0.7, 0])
        vector_obj = Arrow(ORIGIN, initial_vector, buff=0, color=YELLOW)
        vector_label = MathTex(r"\vec{v}").next_to(vector_obj, RIGHT)
        self.play(Create(vector_obj), Write(vector_label))
        self.wait(1)

        matrix = [[2, 1], [1, 1.5]]
        matrix_label = Matrix(matrix).scale(0.7).to_corner(UP + RIGHT)
        self.play(Write(matrix_label))
        self.wait(1)

        transform_scene = MatrixTransformation(matrix, show_basis_vectors=False)
        transform_scene.add(unit_circle, vector_obj, unit_label, vector_label, matrix_label)

        transformed_circle = transform_scene.apply_matrix_to_points(matrix, unit_circle.copy())
        self.play(Transform(unit_circle, transformed_circle), run_time=2)

        transformed_vector = matrix @ initial_vector[:2]
        transformed_arrow = Arrow(ORIGIN, np.append(transformed_vector, 0), color=RED)
        transformed_label = MathTex(r"A \vec{v}").next_to(transformed_arrow, RIGHT)
        self.play(Transform(vector_obj, transformed_arrow), Transform(vector_label, transformed_label))
        self.wait(1)

        amplification_factor = np.linalg.norm(transformed_vector)
        amplification_text = Text(f"Amplification factor ≈ {amplification_factor:.2f}", font_size=24)
        amplification_text.next_to(transformed_arrow, DOWN)
        self.play(Write(amplification_text))
        self.wait(2)

# Use the %%manim cell magic to render inline
%manim -v WARNING -qm L2NormInducedScene
```
